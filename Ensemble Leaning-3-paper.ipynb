```python
for col in data.columns:
    if data[col].dtype == 'float64':
        data[col] = data[col].astype('float32')
    if data[col].dtype == 'int64':
        data[col] = data[col].astype('int32')
    if data[col].dtype == 'object':
        data[col] = data[col].astype('category')
        
```

```python
import numpy as np
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import f1_score
from sklearn import metrics
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import KFold, cross_validate
```

```python
X = data.iloc[:,0:-23].values
Y = data.iloc[:,24].values.ravel()
```

```python
from sklearn.model_selection import train_test_split
X, X_, Y, Y_ = train_test_split(X, Y, test_size=0.3, random_state=0)
```

```python
dt = DecisionTreeClassifier(max_depth=5, min_samples_split=10, random_state=42)
```

```python
estimators = []
#Decision Tree
# dt = DecisionTreeClassifier()
dt.fit(X,Y)
dt_pred =dt.predict(X_)
estimators.append(('Decision Tree', dt))
#Naive Bayes
nb = GaussianNB()
nb.fit(X,Y)
nb_pred =nb.predict(X_)
estimators.append(('Naive Bayes',nb))
```

```python
## BAGGING DECISION TREE
```

```python
ensemble1 = BaggingClassifier(estimator=dt, n_jobs=8, random_state=0)
ensemble1.fit(X,Y)
ensemble1_pred = ensemble1.predict(X_)
#prediksi data frame
hai = {
    'Y True' : Y_,
    'Decision Tree' : dt_pred,
    'Bagging DT' : ensemble1_pred
}
print("PREDIKSI ALGORITMA DECISION TREE DENGAN BAGGING")
pd.DataFrame(hai).head(100)
```

```python
from sklearn.metrics import make_scorer, recall_score

kfold = KFold(n_splits=10)
scoring = {
    'accuracy': 'accuracy',
    'precision_macro': make_scorer(precision_score, average='macro'),
    'recall_macro': make_scorer(recall_score, average='macro'),
    'f1_macro': make_scorer(f1_score, average='macro')
}

# Melakukan cross-validation
hasil = cross_validate(ensemble1, X_, Y_, cv=kfold, scoring=scoring, return_train_score=True)
hasil_ = pd.DataFrame(hasil)

print(hasil_)
```

### Visualiasi Compare Peformance


```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

# Fungsi untuk menampilkan Confusion Matrix sebagai heatmap
def plot_confusion_matrix(y_true, y_pred, title, labels):
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
    plt.title(f'Confusion Matrix: {title}')
    plt.xlabel('Predicted Labels')
    plt.ylabel('True Labels')
    plt.show()

# Label target
labels = ['mirai', 'gafgyt', 'Normal']






```

```python
# Bagging Decision Tree Classification Report dan Confusion Matrix
print("\nClassification Report untuk Bagging Decision Tree")
print(classification_report(Y_, ensemble1_pred, target_names=['mirai', 'gafgyt', 'Normal']))
print("Confusion Matrix untuk Bagging Decision Tree")
# Confusion Matrix untuk Bagging Decision Tree
plot_confusion_matrix(Y_, ensemble1_pred, 'Bagging Decision Tree', labels)
```

```python
# Boosting Decision Tree Classification Report dan Confusion Matrix
print("\nClassification Report untuk Boosting Decision Tree")
print(classification_report(Y_, y_pred_boosting, target_names=['mirai', 'gafgyt', 'Normal']))
print("Confusion Matrix untuk Boosting Decision Tree")
# Confusion Matrix untuk Boosting Decision Tree
plot_confusion_matrix(Y_, y_pred_boosting, 'Boosting Decision Tree', labels)
```

```python
# Stacking Model Classification Report dan Confusion Matrix
print("\nClassification Report untuk Stacking Model")
print(classification_report(Y_, Y_pred_stacking, target_names=['mirai', 'gafgyt', 'Normal']))
print("Confusion Matrix untuk Stacking Model")
# Confusion Matrix untuk Stacking Model
plot_confusion_matrix(Y_, Y_pred_stacking, 'Stacking Model', labels)
```
